# Fyte: 01 - Exploratory Data Analysis (EDA)
## Project: AI-Based Expense Prediction System

In this notebook, we perform comprehensive Exploratory Data Analysis on the primary dataset `expense_data.csv`.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('../../data/raw/expense_data.csv')
print(f"Dataset shape: {df.shape}")
df.head(10)
```

### 1. Dataset Overview and Structural Diagnostics


```python
print(df.info())
print("\nMissing Values:\n", df.isnull().sum())
print("\nDuplicate Rows:", df.duplicated().sum())
print("\nSummary Statistics:\n", df.describe())
```

### 2. Category and Payment Method Distributions


```python
print("Unique Categories:", df['Category'].value_counts())
print("\nPayment Methods:", df['Payment_Method'].value_counts())
```

### 3. Visualizations


```python
plt.figure(figsize=(10, 4))
df['Date'] = pd.to_datetime(df['Date'])
df.sort_values('Date', inplace=True)
plt.plot(df['Date'], df['Amount'], marker='o', color='#2563eb')
plt.title('Daily Expenses Timeline')
plt.xlabel('Date')
plt.ylabel('Amount (INR)')
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.boxplot(x='Category', y='Amount', data=df, palette='Set2')
plt.title('Expense Distribution by Category')
plt.xticks(rotation=30)
plt.show()
```
